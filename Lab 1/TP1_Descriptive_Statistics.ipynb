## Example-1: Find the mean, median, minimum, maximum values, standard deviation as well as the quantiles for salary in USD


### Find the mean, median, min, max value for salary in USD


```python
mean_salary = df['salary_in_usd'].mean()
median_salary = df['salary_in_usd'].median()
min_salary = df['salary_in_usd'].min()
max_salary = df['salary_in_usd'].max()

print(
    "mean: {:.0f}\n"
    "median: {:.0f}\n"
    "min: {:.0f}\n"
    "max: {:.0f}".format(mean_salary, median_salary, min_salary, max_salary)
)
```

### Find the standard deviation and variance for salary in USD

#### The best-known estimates of variability are the *variance* and the *standard deviation*. 

#### Variance  
Variance is a measure of how far the values in a dataset are spread out from their mean. It is calculated as the average of the squared differences between each data point and the mean. Squaring the deviations ensures that all differences are positive and gives more weight to larger deviations.  

$$ 
\large
\text{Variance} \quad = \quad s^2 \quad = \quad \frac{\sum_{i=1}^{n} (x_i - \overline{x})^2}{n - 1}
$$  

#### Standard Deviation  
Standard deviation is the square root of the variance, which brings the units back to the same scale as the data. A larger standard deviation indicates that the data points are more spread out from the mean.  

$$ 
\large
\text{Standard deviation} \quad  s \quad = \quad \sqrt{\text{Variance}}
$$


```python
salary_std = df['salary_in_usd'].std() 
salary_var = df['salary_in_usd'].var()

print(
    "mean: {:.0f}\n"
    "variance: {:.0f}\n"
    "standard deviation: {:.0f}".format(mean_salary, salary_var, salary_std)
)
```

## Example-3: plot the histogram of salary in USD, using 10 bins and with the count displayed on top of each bar
#### Histograms are used to display the distribution of observed data. The data is divided into intervals (or "bins"), and the frequency (or count) of data points within each interval is represented by the height of the bars.

```python
plt.figure(figsize=(8, 5))

# Plot the histogram of salary in USD using 10 bins, with the edge of each bin in black
counts, bins, patches = plt.hist(df['salary_in_usd'], bins=20, edgecolor='black')

# Display the counts on top of each bar
for count, patch in zip(counts, patches):
    # Get the x and y position of the top of each bar
    x = patch.get_x() + patch.get_width() / 2
    y = patch.get_height()
    # Place the count on top of each bar
    plt.text(x, y, int(count), ha='center', va='bottom')


# Use the bin edges as the xticks
plt.xticks(bins, rotation=45)
plt.xlabel("Salary in USD")
plt.ylabel("Count")
plt.title("Histogram of salary in USD")
plt.show()
```

As we can see, the majority of the salary falls within the range **94,000** and **180,000**,  with a count of **1043 + 963 = 2006**.  Given we have a total of **3755 data points**, this indicates about 2006/3755 = **53.4%** of data falling within this range


#### Add vertical lines corresponding to the Q1, Median, Q3 on the histogram

```python
# Quartiles
q1, median, q3 = quantile_25th, quantile_50th, quantile_75th 

plt.figure(figsize=(8, 5))

# Plot histogram
counts, bins, patches = plt.hist(df['salary_in_usd'], bins=20, edgecolor='black')

# Display counts on top of bars
for count, patch in zip(counts, patches):
    x = patch.get_x() + patch.get_width() / 2
    y = patch.get_height()
    plt.text(x, y, int(count), ha='center', va='bottom')

# Overlay quartile lines
plt.axvline(q1, color='green', linestyle='--', linewidth=2, label=f"Q1 = {q1:,}")
plt.axvline(median, color='red', linestyle='-', linewidth=2, label=f"Median = {median:,}")
plt.axvline(q3, color='blue', linestyle='--', linewidth=2, label=f"Q3 = {q3:,}")

# Labels & title
plt.xticks(bins, rotation=45)
plt.xlabel("Salary in USD")
plt.ylabel("Count")
plt.title("Histogram of salary in USD with Quartiles")
plt.legend()
plt.show()
```

As previously mentioned, about half of the dataset (53.4%) falls between ~94,000 and ~180,000. This is quite closely aligned with the IQR range identified with Q1 and Q3.  

#### Run the following code and play with different bin size

```python
from IPython.display import display, HTML

# Side-by-side interactive histogram with bin size legend
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- Widgets: vertical bin choices ---
bin_choices = widgets.RadioButtons(
    options=[10, 30, 100],
    value=10,
    layout=widgets.Layout(width='200px')
)
controls = widgets.VBox([widgets.HTML("<b>Select bin size:</b>"), bin_choices])

# --- Output area for the plot ---
out = widgets.Output()
out.layout = widgets.Layout(width='800px')

def draw_hist(bins):
    with out:
        clear_output(wait=True)
        data = df['salary_in_usd'].dropna().values

        plt.figure(figsize=(8, 5))
        # Add a label to use in the legend
        counts, edges, patches = plt.hist(
            data, bins=bins, edgecolor='black', label=f"{bins} bins"
        )

        # Optional: counts on bars if bins are few
        if bins <= 30:
            for count, patch in zip(counts, patches):
                if count > 0:
                    x = patch.get_x() + patch.get_width() / 2
                    y = patch.get_height()
                    plt.text(x, y, f"{int(count)}", ha='center', va='bottom', fontsize=9)

        # Manage x-ticks
        max_ticks = 12
        if len(edges) <= max_ticks:
            xticks = edges
        else:
            step = max(1, len(edges) // max_ticks)
            xticks = edges[::step]
            if xticks[-1] != edges[-1]:
                xticks = np.append(xticks, edges[-1])

        plt.xticks(xticks, rotation=45)
        plt.xlabel("Salary in USD")
        plt.ylabel("Count")
        plt.title("Histogram of salary in USD")

        # Show legend with current bin size
        plt.legend(loc="upper right")

        plt.tight_layout()
        plt.show()

def on_bins_change(change):
    if change['name'] == 'value':
        draw_hist(change['new'])

bin_choices.observe(on_bins_change, names='value')

# Initial render
draw_hist(bin_choices.value)

# Display layout
ui = widgets.HBox([out, controls])
display(ui)
```

## Example-4: generate the cumulative distribution of salary in USD for employees at SE (senior) level and MI (Mid-level), and show them in a single figure
#### The Seaborn *ecdfplot()* function can be used to plot the Empirical Cumulative Distribution Function(ECDF) of a dataset. The ECDF plot shows the proportion of data $x_i$ that fall below a certain value $x$ in a dataset.
$$
\large F(x) = \frac{1}{n} \sum_{i=1}^{n} 1_{\{x_i \leq x\}}
$$


```python
# Filter the dataset for SE (Senior) and MI (Mid-level) experience levels
df_se = df[df['experience_level'] == 'SE']
df_mi = df[df['experience_level'] == 'MI']

# Create the ECDF plot
plt.figure(figsize=(10, 6))

# Plot the cumulative distribution for SE (Senior) experience level
sns.ecdfplot(data=df_se, x='salary_in_usd', label='SE (Senior)', color='blue')

# Plot the cumulative distribution for MI (Mid-level) experience level
sns.ecdfplot(data=df_mi, x='salary_in_usd', label='MI (Mid-level)', color='green')

# Add labels and title
plt.title('Cumulative Distribution of Salary in USD for SE and MI Experience Levels')
plt.xlabel('Salary in USD')
plt.ylabel('Cumulative Probability')

# Show the legend
plt.legend(title='Experience Level')

# Show the plot
plt.show()
```

## Example-5: create the boxplot for salary in USD. 

### A boxplot provides insightful information about the distribution of the data, such as the median, the 25th quartile, the 75th quartile, IQR, etc. The IQR gives a measure of spread for the middle half data and help understand the central tendency.

#### Execute the following code and generate the boxplot for salary in USD

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(4, 6))

# Create a boxplot for 'salary_in_usd' with vertical orientation
sns.boxplot(y=df['salary_in_usd'], width=0.6)

# Calculate the quantiles
min_val = df['salary_in_usd'].min()
q1_val = df['salary_in_usd'].quantile(0.25)
median_val = df['salary_in_usd'].median()
q3_val = df['salary_in_usd'].quantile(0.75)
iqr = q3_val - q1_val
whisker_min = q1_val - 1.5 * iqr
whisker_max = q3_val + 1.5 * iqr
max_val = df['salary_in_usd'].max()

# Annotate directly at the correct y-positions
plt.text(x=0.15, y=min_val,     s=f'Min = {min_val:.0f}', ha='left', va='bottom', color='blue', fontsize=9)
plt.text(x=0.15, y=whisker_min, s=f'Q1 - 1.5*IQR = {whisker_min:.0f}', ha='left', va='bottom', color='red', fontsize=9)
plt.text(x=0.32, y=q1_val,      s=f'Q1 = {q1_val:.0f}', ha='left', va='bottom', color='blue', fontsize=9)
plt.text(x=0.32, y=median_val,  s=f'Median = {median_val:.0f}', ha='left', va='bottom', color='blue', fontsize=9)
plt.text(x=0.32, y=q3_val,      s=f'Q3 = {q3_val:.0f}', ha='left', va='bottom', color='blue', fontsize=9)
plt.text(x=0.15, y=whisker_max, s=f'Q3 +1.5*IQR = {whisker_max:.0f}', ha='left', va='bottom', color='red', fontsize=9)
plt.text(x=0.15, y=max_val,     s=f'Max = {max_val:.0f}', ha='left', va='bottom', color='blue', fontsize=9)

plt.ylabel("Salary in USD")
plt.title("Boxplot for salary in USD")
plt.show()
```

 ⚠️ Note: **each whisker's end goes to the farthest non-outlier point**. When computed with Q1-1.5*IQR, the theoretical "minimum" is -25000. However, the fathest non-outlier point, or in other words the minimum value of the dataset is 5132. Hence, it is used as the end of the whisker.

## Example-10: Importance of data visualization

#### The **Datasaurus Dozen** is a collection of 13 different datasets that all share nearly the same summary statistics (mean, standard deviation, etc.), but their scatterplots reveal very different shapes and patterns.  It is commonly used to demonstrate the importance of visualization.

**Task:** Run the code below.

```python
# Load the datasaurus dataset
datasaurus = pd.read_csv('DatasaurusDozen.tsv', sep='\t')

# Remove grid, keep background white for cleaner plots.
sns.set(style="white")

# Plot the 13 datasets
# - col="dataset": each dataset gets its own subplot
# - col_wrap=5: wrap plots into rows of max 5 plots per row
# - hue="dataset": color each dataset consistently
# - sharex=False, sharey=False: allow each subplot to have its own x- and y-axis range
# - height=3: each subplot will be 3 inches tall
g = sns.FacetGrid(datasaurus, col="dataset", col_wrap=5, hue="dataset",
                  sharex=False, sharey=False, height=3)

# Map scatterplot onto each facet (small subplot)
# - s=15 sets the marker size
# - alpha=0.7 adds transparency so overlapping points are easier to see
g.map_dataframe(sns.scatterplot, x="x", y="y", s=15, alpha=0.7)
g.set_axis_labels("x", "y")

# Update titles to include dataset name, mean and std.
for ax, dataset_name in zip(g.axes.flat, datasaurus['dataset'].unique()):
    subset = datasaurus[datasaurus['dataset'] == dataset_name]
    mean_x = subset['x'].mean()
    mean_y = subset['y'].mean()
    std_x = subset['x'].std()
    std_y = subset['y'].std()

    ax.set_title(
        f"{dataset_name}\n"
        f"Mean: ({mean_x:.2f}, {mean_y:.2f})\n"
        f"Std: ({std_x:.2f}, {std_y:.2f})\n",
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

For each subplot above, the title includes:  
- The **dataset name** (e.g., `dino`, `circle`, `star`, etc.).  
- The **mean** of the x and y values.  
- The **standard deviation** of the x and y values.

### <span style="color:blue"> [2.5 points] </span> Question-3: create a historgram of salary in USD for full time employee resident in US. The plot should contain 10 bins. 

```python
from IPython.display import display, HTML

# Side-by-side interactive histogram with bin size legend
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- Widgets: vertical bin choices ---
bin_choices = widgets.RadioButtons(
    options=[10],
    value=10,
    layout=widgets.Layout(width='200px')
)
controls = widgets.VBox([widgets.HTML("<b>Select bin size:</b>"), bin_choices])

# --- Output area for the plot ---
out = widgets.Output()
out.layout = widgets.Layout(width='800px')

def draw_hist(bins):
    with out:
        clear_output(wait=True)
        data = df['salary_in_usd'].dropna().values

        plt.figure(figsize=(8, 5))
        # Add a label to use in the legend
        counts, edges, patches = plt.hist(
            data, bins=bins, edgecolor='black', label=f"{bins} bins"
        )

        # Optional: counts on bars if bins are few
        if bins <= 30:
            for count, patch in zip(counts, patches):
                if count > 0:
                    x = patch.get_x() + patch.get_width() / 2
                    y = patch.get_height()
                    plt.text(x, y, f"{int(count)}", ha='center', va='bottom', fontsize=9)

        # Manage x-ticks
        max_ticks = 12
        if len(edges) <= max_ticks:
            xticks = edges
        else:
            step = max(1, len(edges) // max_ticks)
            xticks = edges[::step]
            if xticks[-1] != edges[-1]:
                xticks = np.append(xticks, edges[-1])

        plt.xticks(xticks, rotation=45)
        plt.xlabel("Salary in USD")
        plt.ylabel("Count")
        plt.title("Histogram of salary in USD")

        # Show legend with current bin size
        plt.legend(loc="upper right")

        plt.tight_layout()
        plt.show()

def on_bins_change(change):
    if change['name'] == 'value':
        draw_hist(change['new'])

bin_choices.observe(on_bins_change, names='value')

# Initial render
draw_hist(bin_choices.value)

# Display layout
ui = widgets.HBox([out, controls])
display(ui)
```

### <span style="color:blue"> [3.5 points] </span> Question-4: create a boxplot of salary in USD for full time employee at each experience level. 

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Filter dataset for Full-Time employees
df_ft = df[df["employment_type"] == "FT"].copy()

# Set up figure size
plt.figure(figsize=(10, 6))

# Order of experience levels for logical visualization
exp_order = ["EN", "MI", "SE", "EX"]

# 2. Create Boxplot grouped by Experience Level
ax = sns.boxplot(
    data=df_ft,
    x="experience_level",
    y="salary_in_usd",
    order=exp_order,
    palette="Blues",
    width=0.5,
)

# 3. Calculate and annotate key statistics for each experience level
for i, level in enumerate(exp_order):
    sub_data = df_ft[df_ft["experience_level"] == level]["salary_in_usd"]

    if sub_data.empty:
        continue

    q1_val = sub_data.quantile(0.25)
    median_val = sub_data.median()
    q3_val = sub_data.quantile(0.75)
    iqr = q3_val - q1_val
    whisker_min = sub_data[sub_data >= (q1_val - 1.5 * iqr)].min()
    whisker_max = sub_data[sub_data <= (q3_val + 1.5 * iqr)].max()

    # Annotate Median and Quartiles next to each box
    ax.text(
        i + 0.28,
        median_val,
        f"Med: ${median_val:,.0f}",
        ha="left",
        va="center",
        fontsize=8,
        color="darkblue",
        weight="bold",
    )
    ax.text(
        i + 0.28,
        q1_val,
        f"Q1: ${q1_val:,.0f}",
        ha="left",
        va="center",
        fontsize=8,
        color="gray",
    )
    ax.text(
        i + 0.28,
        q3_val,
        f"Q3: ${q3_val:,.0f}",
        ha="left",
        va="center",
        fontsize=8,
        color="gray",
    )

# Formatting
plt.title(
    "Salary (USD) Distribution by Experience Level for Full-Time Employees",
    fontsize=12,
    pad=15,
)
plt.xlabel("Experience Level", fontsize=10)
plt.ylabel("Salary in USD", fontsize=10)
plt.xticks(
    ticks=[0, 1, 2, 3],
    labels=[
        "Entry-Level (EN)",
        "Mid-Level (MI)",
        "Senior-Level (SE)",
        "Executive (EX)",
    ],
)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()

plt.show()
```
